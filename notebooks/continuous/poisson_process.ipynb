```python
import numpy as np
import plotly.graph_objs as go
```

```python
from samplepaths.continuous import PoissonProcess as TestType
from stochastic.continuous import PoissonProcess as StochasticTestType
```

```python
seed = 1234567890
n = 10
rate = 0.5

rng = np.random.default_rng(seed)
np.random.seed(seed)
```

```python
x = TestType(
    rate=rate,
    rng=rng,
)
xs = x.sample(n=n)

y = StochasticTestType(
    rate=rate,
)
ys = y.sample(n=n)
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        y=xs,
        mode="lines",
        name="Current",
    )
)

fig.add_trace(
    go.Scatter(
        y=ys,
        mode="lines",
        name="Original",
    )
)


fig.update_layout(
    title="Samples",
    xaxis_title="Sample Index",
    yaxis_title="Amplitude",
)

fig.show()
```
